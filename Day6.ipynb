```python
def even_odd(x):

    if x < 0:
        print("its negative", x)

    elif x == 0:
        print("its 0")

    elif x % 2 == 0:
        print("its even no", x)

    else:
        print("its odd", x)

even_odd(2)

even_odd(0)

even_odd(-2)

even_odd(3)
```

```python
def fact(x):
    product = 1

    for i in range(1, x + 1):
        product = product * i

    return product

fact(5)
```

```python
def fact(x):

    if x == 1:
        return 1
    else:
      return x * fact(x - 1)


print(fact(5))
```

```python
def simple_interest(p, r, t):
    interest = (p * r * t) / 100

    print(
        "Interest on principal amount", p,
        "at rate", r,
        "for", t, "years is", interest
    )

simple_interest(1000, 2, 2)
```

```python
def calculate(a,b,operation = "add"):
  if operation == "add":
    return a+b
  elif operation=="mul":
    return a*b
  elif operation=="sub":
    return a-b
  elif operation=="div":
    return a/b

calculate(12,13)
```

```python
def is_prime(n):
    if n <= 1:
        return False

    count = 0

    for i in range(1, n + 1):
        if n % i == 0:
            count = count + 1


    return count == 2

print(is_prime(4))

```

```python
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n - 1) + fib(n - 2)


n = int(input("Enter n: "))
print(" { " ,end="")
for i in range(n):
    print(fib(i), end=" ")
print("}")
```

```python
points = [ (8,9),(1,2) , (1,3), (4,5),(8,1)]
points.sort()
print(points)
```

```python
points = [ ('a',90),('b',20) , ('c',39), ('d',54),('a',15)]
points.sort()
print(points)
```

```python
points = [ ('a',90),('b',20) , ('c',39), ('d',54),('a',15)]
points.sort(key = lambda  p: p[1])
print(points)
```

```python
nums = [1,2,3,4,5,6]
even = filter(lambda x:x % 2 == 0,nums)
print(list(even))

```

```python
nums = [1, 2, 3, 4, 5, 6]

square = map(lambda x: x * x, nums)

print(list(square))
```
