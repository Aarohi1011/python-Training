**Prime no or not**

```python
for n in range(2, 101):
    count = 0

    for i in range(1, n + 1):
        if n % i == 0:
            count = count + 1

    if count == 2:
        print(n, "is prime")
```

**fibonnaci**

```python
o = int(input("Enter number of terms: "))

first = 0
second = 1

for i in range(o):
    print(first)

    next_number = first + second
    first = second
    second = next_number
```

**Armstrong no for 3 digit**

```python
num = int(input("Enter 3 digit number: "))

hundred = num // 100
ten = (num // 10) % 10
one = num % 10

if hundred**3 + ten**3 + one**3 == num:
    print("Armstrong number")
else:
    print("Not Armstrong number")
```

**Armstrong for any no**

```python
num5 = int(input("Enter number to check: "))

total1 = 0
original = num5

digitslen = len(str(num5))

while num5 > 0:

    digit = num5 % 10
    total1 = total1 + digit ** digitslen
    num5 = num5 // 10

if total1 == original:
    print("Its Armstrong")
else:
    print("Not Armstrong")
```
