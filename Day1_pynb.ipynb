```python
greet = "hello, world"

print("First character :", greet[0])
print("Last character  :", greet[-1])
print("Second character:", greet[1])
print("First 5 chars   :", greet[:5])
print("World           :", greet[7:])
print("Length          :", len(greet))
print("Uppercase       :", greet.upper())
print("Title           :", greet.title())
print("Reverse         :", greet[::-1])
print("Find world      :", greet.find("world"))
print("Replace         :", greet.replace("world", "Aarohi"))
print("Split           :", greet.split())
print("Strip           :", greet.strip())
print("World present?  :", "world" in greet)
```

# **Restraunt Bill Generator**


```python
from re import sub

print("🍽️ RESTAURANT MENU")


print("1. Pizza          ₹200")
print("2. Burger         ₹120")
print("3. Pasta          ₹150")
print("4. Sandwich       ₹100")
print("5. Cold Drink      ₹50")
print("6. Coffee          ₹80")



choice = int(input("Enter item number: "))
quantity = int(input("Enter quantity: "))

if choice == 1:
    item = "Pizza"
    price = 200

elif choice == 2:
    item = "Burger"
    price = 120

elif choice == 3:
    item = "Pasta"
    price = 150

elif choice == 4:
    item = "Sandwich"
    price = 100

elif choice == 5:
    item = "Cold Drink"
    price = 50

elif choice == 6:
    item = "Coffee"
    price = 80

else:
    print("❌ Invalid item number!")
    exit()

subtotal = price * quantity
gst_rate = 18
gst_amount = subtotal * gst_rate / 100


grand_total = subtotal + gst_amount



print("🧾 BILL")


print("Item       :", item)
print("Price      : ₹", price)
print("Quantity   :", quantity)
print("GST        :",  gst_amount)
print("Total      : ₹", grand_total)



print("Thank You! Visit Again 😊")
```
