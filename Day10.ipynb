```python
import pandas as pd

df = pd.read_csv("/content/data.csv")

print("=" * 50)
print("             STUDENT DATA")
print("=" * 50)

print("\n📋 COMPLETE DATA")
print(df.to_string(index=True))

print("\n" + "=" * 50)
print("             BASIC INFORMATION")
print("=" * 50)

print(f"Total Rows    : {df.shape[0]}")
print(f"Total Columns : {df.shape[1]}")

print("\nColumn Names:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

print("\nData Types:")
print(df.dtypes)

print("\n" + "=" * 50)
print("             STATISTICS")
print("=" * 50)

print(f"Average Age   : {df['Age'].mean():.2f}")
print(f"Minimum Age   : {df['Age'].min()}")
print(f"Maximum Age   : {df['Age'].max()}")
print(f"Total Age     : {df['Age'].sum()}")
print(f"Age Count     : {df['Age'].count()}")

print("\n" + "=" * 50)
print("             DATAFRAME INFO")
print("=" * 50)

df.info()
```

```python
import pandas as pd

data = {
    "name": ["Asha", "Lata", "Meena", "Sneha"],
    "age": [21, 26, 27, 25],
    "city": ["Bhopal", "Indore", "Mumbai", "Goa"]
}

df = pd.DataFrame(data)
print(df["name"])
print(df[["name","age"]])
df["age_next"] = df["age"]+1
print(df)

print(f"Average Age   : {df['age'].mean():.2f}")

```

```python
import pandas as pd

data = {
    "name": ["Asha", "Lata", "Meena", "Sneha"],
    "age": [21, 26, 27, 25],
    "city": ["Bhopal", "Indore", "Mumbai", "Goa"]
}
df = pd.DataFrame(data)
print(df.loc[0])
print(df.loc[0:2,"name"])
print(df.loc[:,["name","city"]])
print(df.iloc[0])
print(df.iloc[0:2,0:2])
print(df.iloc[-1])
```
