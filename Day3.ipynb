**Vowel and consonants**

```python
c = "Neeta"
count = 0
length = len(c)
vowel = "aeiouAEIOU"
for i in c:
  if i in vowel:
    count= count+1
print("vowel" , count , "consonant" , (length-count))
```

## **Matrix printing**

```python
matrix = [[1,2,3,74],
         [4,5,6,16],
         [7,8,9,89],
         [10,11,12,45]]
print(matrix)

for row in matrix:
  for val in row:
    print(val, end = " ")

  print()
```
