```python
from collections import defaultdict
words = ["apple","cherry","mango","avacado","cherry", "cashew","melon"]

groups = defaultdict(list)

for w in words:
  groups[w[0]].append(w)
print(dict(groups))
```

```python
from collections import deque
dq = deque([2,3,4])

dq.append(5)
print(dq)


dq.appendleft(1)
print(dq)

dq.rotate(2)
print(dq)

window = deque(maxlen=2)
for x in [1,2,3,4]:
  window.append(x)
  print(list(window))
```
