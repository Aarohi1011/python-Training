# **Set**

```python
s1 = { 1,2,8,9,15,18,1,2,5,12}
print(s1)

s1.add(54)
print(s1)

s1.update((79,56))
print(s1)

s1.remove(54)
print(s1)

s1.discard(54)
print(s1)

s1.pop()
print(s1)

print(len(s1))
```

```python
a = {1,2,3,9,6}
b = {1,4,2,9,10,45}
print(a.union(b))
print( a|b )

print(a.intersection(b))
print( a&b )

a.intersection_update(b)
print(a)


```

```python
a = {1,2,3,4}
b = {9,10}
print(a.difference(b))
print(a.symmetric_difference(b))
print({5}.issubset(a))
print(a.isdisjoint(b))
```

# **Dictionary**

```python
student = {
     "name" : "Neha",
     "age"   : 22,
     "course" : "Python"
 }

print(student["name"])

print(student.get("name"))

print(student.get("city","NA"))

student["city"] = "Bhopal"
print(student.get("city"))

print(student.keys())

print(student.values())

print(student.items())

student.update({"age" : 20})
print(student.get("age"))

student.pop("age")
print(student)

student.popitem()
print(student)


```

```python
n = int(input("enter no values we store in dictonaries "))
dict = {}
for i in range(1, n + 1):
  key = input("enter value of key ")
  value = input("enter value of key ")
  dict[key] = value
print(dict)
```
