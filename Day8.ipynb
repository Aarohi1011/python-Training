```python
import random

a = random.random()
print("Random decimal between 0 and 1:", a)

b = random.randint(1, 69)
print("Random integer between 1 and 69:", b)

c = random.randrange(1, 9)
print("Random integer between 1 and 8:", c)

items = ["a", "b", 12, "Aarohi", "deck", 1, 52]

d = random.choice(items)
print("One random item from the list:", d)

e = random.choices(items, k=3)
print("Three random items, repetition is possible:", e)

f = random.sample(items, 3)
print("Three random items, no repetition:", f)

g = random.uniform(1, 10)
print("Random decimal between 1 and 10:", g)
```

**Exception Handling**

```python
try:
  num = int(input("Enter any number "))
  print(10/num)
except ValueError:
  print("Please enter a Valid number")
except ZeroDivisionError:
  print("Cannot divide by zero")
```

```python
try:
  num = int(input("Enter any number "))
  print(10/num)

except (ValueError, TypeError) as e:
    print("Input Problem:-", e)

except Exception as e:
    print("Unexpected Error:-", e)
```

```python
try:
    a = int(input("Enter a no:- "))

    result = 10 / a

except ZeroDivisionError:
    print("Error!")

else:
    print("Success:", result)

finally:
    print("Execution finished")
```
